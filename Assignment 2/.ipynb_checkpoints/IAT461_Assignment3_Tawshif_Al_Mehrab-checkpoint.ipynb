## Part A Business-Level Clustering
### A1 Data Acquisition and Cleaning
This section loads and examines the City of Vancouver Business Licences
GeoJSON dataset. Cleaning decisions will be based on missingness, geographic
availability, licence status, and the suitability of each field for clustering.

```python
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# AI assistance
# Tool: Peplexity AI
# Prompt: "Help me load and initially inspect the Vancouver Business
warnings.filterwarnings("ignore")
FILE_PATH = "business-licences.geojson"
try:
    with open(FILE_PATH, "r", encoding="utf-8") as file:
        geojson_data = json.load(file)

    print("GeoJSON loaded successfully.")
    print("Number of features:", len(geojson_data["features"]))

except FileNotFoundError:
    print(
        f"File not found: {FILE_PATH}\n"
        "Place the GeoJSON file in the same folder as this notebook.")
```

```python
records = []
for feature in geojson_data["features"]:
    properties = feature.get("properties", {}).copy()
    geometry = feature.get("geometry")
    longitude = np.nan
    latitude = np.nan
    if geometry and geometry.get("type") == "Point":
        coordinates = geometry.get("coordinates", [])
        if len(coordinates) >= 2:
            longitude = coordinates[0]
            latitude = coordinates[1]
    properties["longitude"] = longitude
    properties["latitude"] = latitude
    records.append(properties)
df = pd.DataFrame(records)
print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Columns:")
print(df.columns.tolist())
print("\nData types:")
display(df.dtypes.to_frame("data_type"))
duplicate_columns = [column
    for column in df.columns
    if column != "geo_point_2d"]
print("\nDuplicate rows:",df.duplicated(subset=duplicate_columns).sum())
missing_summary = ( df.isna().mean().mul(100).sort_values(ascending=False).to_frame("missing_percentage"))
display(missing_summary.head(20))
important_missing = ["numberofemployees","feepaid","businesssubtype","postalcode","latitude","longitude"]
display(missing_summary.loc[important_missing])
```

```python
print("Status counts:")
display(df["status"].value_counts(dropna=False))
print("\nBusiness type counts:")
display(df["businesstype"].value_counts().head(20))
print("\nNumber of employees summary:")
display(df["numberofemployees"].describe())
print("Percentage with zero employees:",(df["numberofemployees"] == 0).mean() * 100)
print("\nFee paid summary:")
display(df["feepaid"].describe())
print("\nDate examples:")
display(df[["issueddate", "expireddate"]].head(10))
```

```python
df_clean = df.drop(columns=["geom", "geo_point_2d"],errors="ignore").copy()
df_clean["issueddate"] = pd.to_datetime(df_clean["issueddate"],errors="coerce",utc=True)
df_clean["expireddate"] = pd.to_datetime(df_clean["expireddate"],errors="coerce",utc=True)
df_clean["licencerevisionnumber"] = pd.to_numeric(df_clean["licencerevisionnumber"],errors="coerce")
df_clean["numberofemployees"] = pd.to_numeric(df_clean["numberofemployees"],errors="coerce")
df_clean["feepaid"] = pd.to_numeric(df_clean["feepaid"],errors="coerce")
print("Cleaned shape:", df_clean.shape)
display(df_clean.head())
```

```python
df_issued = df_clean[df_clean["status"] == "Issued"].copy()
print("Issued licences:", df_issued.shape)
```

```python
df_location = df_issued.dropna(subset=["latitude", "longitude"]).copy()
print("Location dataset shape:", df_location.shape)
display(
    df_location[
        ["licencenumber", "businesstype", "localarea",
         "latitude", "longitude"]
    ].head())
```

```python
display(
    df_location[ ["latitude", "longitude"]].describe())
```

```python
df_location = df_location[df_location["latitude"].between(49.0, 49.5) &df_location["longitude"].between(-123.5, -122.8)].copy()
print("After coordinate validation:", df_location.shape)
```

## A2 Location-Only Clustering

this clusters businesses using only latitude and longitude. Two algorithms were applied-
1. k-mean clustering
2. DBSCAN
the purpose is to compare a partition-based algorithm with a density-based algorithm and examine how their assumptions affect the resulting geographic groupings

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
location_features = df_location[["latitude", "longitude"]].copy()
scaler_location = StandardScaler()
location_scaled = scaler_location.fit_transform(
    location_features)
print("Scaled location shape:", location_scaled.shape)
```

```python
location_sample = df_location.sample(
    n=min(20000, len(df_location)),
    random_state=42
)
sample_scaled = scaler_location.fit_transform(
    location_sample[["latitude", "longitude"]]
)
```

```python
inertias = []
k_values = range(2, 11)
for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(sample_scaled)
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(list(k_values),inertias,marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Location-Based K-Means")
plt.xticks(list(k_values))
plt.grid(alpha=0.3)
plt.show()
```

```python
chosen_k = 5
location_scaled = scaler_location.fit_transform(df_location[["latitude", "longitude"]])
kmeans_location = KMeans(n_clusters=chosen_k,random_state=42,n_init=10)
df_location["kmeans_cluster"] = (kmeans_location.fit_predict(location_scaled))

display(df_location[
        ["latitude", "longitude", "localarea", "kmeans_cluster"]
    ].head())
print(df_location["kmeans_cluster"].value_counts().sort_index())
```

```python
plt.figure(figsize=(10, 8))
scatter = plt.scatter(df_location["longitude"],df_location["latitude"],c=df_location["kmeans_cluster"],s=6,alpha=0.8,cmap="Set2",edgecolor="k",linewidth=0.2)
plt.xlabel("longitude")
plt.ylabel("latitude")
plt.title(f"kmeans location clustering Vancouver busineses "f"(K={chosen_k})")

plt.colorbar(scatter,label="Cluster")
plt.show()
```

**A2 interpretation**
Elbow method showed a noticeable decrease in inertia until approximately K=5, after which the improvement became much smaller. Therefore, five clusters were selected for the kMeans model. The resulting clusters divide vancouver into distinct geographic regions that correspond to different areas of the city. The largest cluster contains the dense central business districts, while smaller clusters represent surrounding regions with fewer businesses. This suggests that business licences naturally group according to geographic proximity.

```python
from sklearn.cluster import DBSCAN
```

```python
eps_values = [0.05, 0.08, 0.10, 0.12, 0.15]
```

```python
eps_values = [0.05, 0.08, 0.10, 0.12, 0.15]
for eps in eps_values:

    dbscan = DBSCAN(
        eps=eps,
        min_samples=25
    )

    labels = dbscan.fit_predict(location_scaled)

    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

    noise = (labels == -1).sum()

    print(
        f"eps={eps:.2f} | "
        f"clusters={n_clusters} | "
        f"noise={noise}"
    )
```

```python
dbscan = DBSCAN(
    eps=0.10,
    min_samples=25
)

df_location["dbscan_cluster"] = dbscan.fit_predict(
    location_scaled
)
```

```python
print(
    df_location["dbscan_cluster"]
    .value_counts()
    .sort_index()
)

print(
    "Noise points:",
    (df_location["dbscan_cluster"] == -1).sum()
)
```

```python
noise = df_location[df_location["dbscan_cluster"] == -1]
clusters = df_location[df_location["dbscan_cluster"] != -1]

plt.figure(figsize=(10, 8))

plt.scatter(
    clusters["longitude"],
    clusters["latitude"],
    c=clusters["dbscan_cluster"],
    cmap="tab20",
    s=6,
    alpha=0.7
)

plt.scatter(
    noise["longitude"],
    noise["latitude"],
    c="black",
    s=10,
    label="Noise / Outliers"
)

plt.title("DBSCAN Clustering of Vancouver Business Locations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.legend()
plt.tight_layout()
plt.show()
```

### DBSCAN Results

DBSCAN produced one large connected cluster, several smaller local clusters, and a set of noise points.

**Interpretation:**  
DBSCAN identified one large connected cluster, several smaller local clusters, and a number of noise points, the noise points represent businesses that were too isolated to belong to a dense group. The large cluster likely formed because many Vancouver businesses are located along continuous commercial corridors, allowing nearby points to connect across a wide area. The smaller clusters may represent more isolated neighbourhood business centres or concentrated commercial zones.

## A3 Feature-Based Clustering= Size, Industry, Lifecycle

this analysis groups businesses using operational characteristics rather than geographic location-
- log-transformed number of employees
- log-transformed licence fee
- licence duration in days
- issue month encoded using sine and cosine
- licence revision number
- consolidated business type.

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

df_a3 = df_issued.copy()
df_a3["licence_duration_days"] = (df_a3["expireddate"] - df_a3["issueddate"]).dt.days
df_a3["issue_month"] = df_a3["issueddate"].dt.month
df_a3["issue_month_sin"] = np.sin(2 * np.pi * df_a3["issue_month"] / 12)
df_a3["issue_month_cos"] = np.cos(2 * np.pi * df_a3["issue_month"] / 12)
```

```python
a3_columns = ["employees_log","fee_log","licence_duration_days","issue_month_sin","issue_month_cos","licencerevisionnumber","business_type_grouped"]
df_a3_model = df_a3[a3_columns].dropna().copy()
print("A3 dataset shape:", df_a3_model.shape)
display(df_a3_model.head())
```

```python
numeric_features = ["employees_log","fee_log","licence_duration_days","issue_month_sin","issue_month_cos","licencerevisionnumber"]
categorical_features = ["business_type_grouped"]
preprocessor = ColumnTransformer(
    transformers=[
        ( "numeric", StandardScaler(),numeric_features),("categorical",OneHotEncoder(handle_unknown="ignore",sparse_output=False),categorical_features)])
X_a3 = preprocessor.fit_transform(df_a3_model)
print("Feature matrix shape:", X_a3.shape)
```

```python
rng = np.random.default_rng(42)
sample_size = min(20000, len(X_a3))
sample_indices = rng.choice(len(X_a3),size=sample_size,replace=False)
X_a3_sample = X_a3[sample_indices]
inertias_a3 = []
k_values_a3 = range(2, 9)
for k in k_values_a3:
    model = KMeans(n_clusters=k,random_state=42,n_init=10)
    model.fit(X_a3_sample)
    inertias_a3.append(model.inertia_)
plt.figure(figsize=(8, 5))
plt.plot(list(k_values_a3),inertias_a3,marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Feature-Based Clustering")
plt.xticks(list(k_values_a3))
plt.grid(alpha=0.3)
plt.show()
```

```python
chosen_k_a3 = 5

kmeans_a3 = KMeans(
    n_clusters=chosen_k_a3,
    random_state=42,
    n_init=10
)

a3_clusters = kmeans_a3.fit_predict(X_a3)

df_a3_model["feature_cluster"] = a3_clusters

print(
    df_a3_model["feature_cluster"]
    .value_counts()
    .sort_index()
)
```

```python
pca = PCA(n_components=2)

X_a3_pca = pca.fit_transform(X_a3)

df_a3_model["PC1"] = X_a3_pca[:, 0]
df_a3_model["PC2"] = X_a3_pca[:, 1]

print(
    "Explained variance:",
    pca.explained_variance_ratio_
)

print(
    "Total explained variance:",
    pca.explained_variance_ratio_.sum()
)
```

```python
pca_plot_sample = df_a3_model.sample(
    n=min(30000, len(df_a3_model)),
    random_state=42
)

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    pca_plot_sample["PC1"],
    pca_plot_sample["PC2"],
    c=pca_plot_sample["feature_cluster"],
    s=6,
    alpha=0.5,
    cmap="tab10"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title(
    "PCA Projection of Feature-Based Business Clusters"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.tight_layout()
plt.show()
```

```python
cluster_profile = (
    df_a3_model
    .groupby("feature_cluster")
    .agg(
        Businesses=("feature_cluster", "size"),
        AvgEmployeesLog=("employees_log", "mean"),
        AvgFeeLog=("fee_log", "mean"),
        AvgLicenceDuration=("licence_duration_days", "mean"),
        AvgRevision=("licencerevisionnumber", "mean")
    )
    .round(2)
)

display(cluster_profile)
```

```python
dominant_types = (
    df_a3_model
    .groupby("feature_cluster")["business_type_grouped"]
    .value_counts()
    .groupby(level=0)
    .head(5)
)

display(dominant_types)
```

```python
df_b1 = df_issued.dropna(
    subset=["localarea", "business_type_grouped"]
).copy()

area_counts = df_b1["localarea"].value_counts()

print("Number of local areas:", area_counts.size)
display(area_counts.sort_values())
```

```python
area_composition = pd.crosstab(
    df_b1["localarea"],
    df_b1["business_type_grouped"],
    normalize="index"
) * 100

print("Composition matrix shape:", area_composition.shape)
display(area_composition.head())
```

```python
area_summary = (
    df_b1.groupby("localarea")
    .agg(
        business_count=("licencenumber", "count"),
        latitude=("latitude", "mean"),
        longitude=("longitude", "mean")
    )
)

display(area_summary.head())
```

```python
valid_areas = area_counts[
    (area_counts >= 500) &
    (area_counts.index != "Out of Town")
].index

df_b1_filtered = df_b1[
    df_b1["localarea"].isin(valid_areas)
].copy()

area_composition = pd.crosstab(
    df_b1_filtered["localarea"],
    df_b1_filtered["business_type_grouped"],
    normalize="index"
) * 100

area_summary = (
    df_b1_filtered.groupby("localarea")
    .agg(
        business_count=("licencenumber", "count"),
        latitude=("latitude", "mean"),
        longitude=("longitude", "mean")
    )
)

print("Areas retained:", len(area_composition))
print("Composition matrix shape:", area_composition.shape)

display(area_composition.head())
display(area_summary.head())
```
